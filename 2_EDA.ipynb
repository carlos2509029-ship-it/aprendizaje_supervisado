Agrupa las edades en categorías (Infante, Niño, Adolescente, Joven Adulto, Adulto, Mayor) y crea una gráfica que muestre la tasa de supervivencia por grupo de edad.

Primero vamos a crear una nueva columna llamada AgeGroup, donde clasificaremos a los pasajeros en seis categorías. Después calcularemos el porcentaje de supervivencia de cada grupo y lo mostraremos en una gráfica.

1. Crear las categorías de edad

```python
import pandas as pd

bins = [0, 2, 12, 17, 29, 59, float('inf')]

labels = [
    'Infante',
    'Niño',
    'Adolescente',
    'Joven Adulto',
    'Adulto',
    'Mayor'
]

df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

Explicación:
- bins: establece los límites de edad de cada categoría.
- labels: asigna el nombre correspondiente a cada intervalo.
- pd.cut(): clasifica cada edad dentro de una categoría.
- df['AgeGroup']: crea una nueva columna en nuestro DataFrame.
Los pasajeros que no tienen registrada su edad (NaN) quedarán sin categoría.

2. Calcular la tasa de supervivencia por grupo

Explicación:
- groupby('AgeGroup'): agrupa los pasajeros por categoría de edad.
- ['Survived']: selecciona la columna que indica si sobrevivieron (1) o no (0).
- .mean(): calcula el promedio de supervivencia de cada grupo.
- * 100: convierte el resultado en porcentaje.
Por ejemplo, si un grupo obtiene 60, significa que sobrevivió el 60 % de los pasajeros de esa categoría.

```python
survival_by_age = (
    df.groupby('AgeGroup', observed=False)['Survived']
    .mean() * 100
)

survival_by_age
```

3. Crear la gráfica

¿Cómo interpretar la gráfica?
- Eje X: muestra las seis categorías de edad.
- Eje Y: muestra el porcentaje de pasajeros que sobrevivieron en cada grupo.
- Altura de las barras: cuanto más alta sea una barra, mayor fue la tasa de supervivencia de ese grupo.
Ejemplo de interpretación: si la categoría Infante tiene una tasa del 60 %, significa que 6 de cada 10 pasajeros de ese grupo sobrevivieron.

La gráfica nos permitirá observar si los pasajeros más jóvenes tuvieron una mayor tasa de supervivencia que los adultos o las personas mayores.

Una observación importante: los porcentajes representan asociaciones en los datos, no demuestran por sí solos que la edad haya sido la causa de la supervivencia.

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=survival_by_age.index,
    y=survival_by_age.values,
    hue=survival_by_age.index,
    palette='pastel',
    legend=False
)

plt.title('Tasa de supervivencia por grupo de edad')
plt.xlabel('Grupo de edad')
plt.ylabel('Tasa de supervivencia (%)')
plt.xticks(rotation=20)
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

Crea una gráfica que muestre la distribución de edades, incluyendo líneas verticales que indiquen la media, mediana y moda de las edades.

Crearemos un histograma de las edades de los pasajeros y agregaremos tres líneas verticales para identificar las medidas de tendencia central: media, mediana y moda.

¿Qué estamos calculando?
- Media (mean()): suma de todas las edades dividida entre el número de pasajeros con edad registrada.
- Mediana (median()): edad que queda en el centro cuando ordenamos todas las edades.
- Moda (mode()): edad que aparece con mayor frecuencia.
Las líneas verticales representan cada una de estas medidas:
Color	Medida
Rojo	Media
Verde	Mediana
Naranja	Moda

Resultado esperado
Si utilizas el Titanic clásico, los valores aproximados serán:
Medida	Edad
Media	29.70 años
Mediana	28 años
Moda	24 años

¿Cómo interpretar la gráfica?
- La media indica que la edad promedio de los pasajeros era de aproximadamente 30 años.
- La mediana indica que la mitad de los pasajeros tenía 28 años o menos y la otra mitad 28 años o más.
- La moda indica que 24 años era la edad que aparecía con mayor frecuencia.
Conclusión: las edades de los pasajeros se concentraban principalmente en adultos jóvenes. La media es ligeramente superior a la mediana, lo que sugiere cierta inclinación de la distribución hacia edades mayores.

Nota: los valores pueden variar ligeramente según los datos y el tratamiento de las edades faltantes en tu CSV.

```python
# Calcular medidas de tendencia central
media = df['Age'].mean()
mediana = df['Age'].median()
moda = df['Age'].mode()[0]

# Crear la gráfica
plt.figure(figsize=(11, 6))

sns.histplot(
    data=df,
    x='Age',
    bins=30,
    kde=True,
    color='steelblue'
)

# Agregar líneas verticales
plt.axvline(media, color='red', linestyle='--',
            linewidth=2, label=f'Media: {media:.2f}')

plt.axvline(mediana, color='green', linestyle='--',
            linewidth=2, label=f'Mediana: {mediana:.2f}')

plt.axvline(moda, color='orange', linestyle='--',
            linewidth=2, label=f'Moda: {moda:.2f}')

# Personalizar gráfica
plt.title('Distribución de edades de los pasajeros del Titanic')
plt.xlabel('Edad de los pasajeros')
plt.ylabel('Cantidad de pasajeros')
plt.legend()

plt.tight_layout()
plt.show()
```

Crea un diagrama de caja (boxplot) que muestre los outliers en la columna de edad.

Crearemos un diagrama de caja para visualizar la distribución de las edades e identificar posibles outliers (valores atípicos).

¿Cómo interpretar la gráfica?
Un diagrama de caja muestra cinco elementos importantes:
- Línea central: representa la mediana de las edades.
- Borde izquierdo de la caja (Q1): indica el 25 % de las edades.
- Borde derecho de la caja (Q3): indica el 75 % de las edades.
- Bigotes: muestran el rango de valores que no se consideran atípicos según el criterio del diagrama.
- Puntos fuera de los bigotes: representan posibles valores atípicos (outliers).

¿Qué resultado esperamos?
En el Titanic clásico, es posible encontrar puntos por encima del bigote superior, correspondientes a pasajeros de edades relativamente elevadas, por ejemplo, personas de 65, 70 o 80 años.

Una observación: como Survived es una variable binaria (0 o 1), la correlación de Pearson puede utilizarse, pero la línea no representa una predicción de probabilidad tan adecuada como una regresión logística.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Crear datos de ejemplo
x = np.array([0.8, 1.7, 2.6, 3.4, 4.3, 5.2, 6.1, 7.0, 8.0, 9.1])
y = np.array([1.4, 2.6, 2.2, 4.1, 3.8, 5.7, 6.3, 6.7, 8.5, 8.1])

# Calcular correlación
correlacion = np.corrcoef(x, y)[0, 1]

# Crear gráfica
plt.figure(figsize=(10, 6))

sns.regplot(
    x=x,
    y=y,
    ci=None,
    color='blue',
    scatter_kws={'s': 70},
    line_kws={'color': 'green', 'linewidth': 2}
)

plt.title(f'Correlación lineal positiva (r = {correlacion:.2f})')
plt.xlabel('X')
plt.ylabel('Y')

plt.tight_layout()
plt.show()
```
